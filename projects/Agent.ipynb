## Step 1: Explore Table Schemas and Sample Data
We need to understand the structure and recent data in both tables. We'll display the schema and sample 1000 rows from each table, filtered by the most recent 'Last Updated' dates if possible.

```python
# Display schema
df_change_orders = spark.read.table('gshen_catalog.enb_projects_workshop.change_orders')
df_change_orders.printSchema()

# Display sample data (last 12 months if possible)
from datetime import date, timedelta
recent_date = date.today() - timedelta(days=365)
df_change_orders_sample = df_change_orders.filter(df_change_orders['Last Updated'] >= recent_date.isoformat()).limit(1000)
display(df_change_orders_sample)
```
