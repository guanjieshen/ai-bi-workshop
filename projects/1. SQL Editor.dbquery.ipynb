```sql
select * from traffic_incidents

CREATE OR REPLACE TABLE traffic_incidents AS 
SELECT 
  id,
  incident_info,
  description,
  start_dt
FROM traffic_incidents

CREATE MATERIALIZED VIEW traffic_incidents_threeyears AS
SELECT
  id
  , incident_info
  , description
  , start_dt
FROM traffic_incidents
WHERE start_dt >= date_add(YEAR, -3, current_date())

INSERT INTO traffic_incidents (id, incident_info, description, start_dt) VALUES
  ('1111', 'Info AAA', 'Description AAA', TIMESTAMP '2024-01-01 00:00:00'),
  ('2222', 'Info BBB', 'Description BBB', TIMESTAMP '2024-01-02 00:00:00'),
  ('3333', 'Info CCC', 'Description CCC', TIMESTAMP '2024-01-03 00:00:00'),
  ('4444', 'Info DDD', 'Description DDD', TIMESTAMP '2021-01-03 00:00:00');

-- refresh materialized view traffic_incidents_threeyears
select * from traffic_incidents_threeyears
```
