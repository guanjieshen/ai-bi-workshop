```sql
-- Databricks supports invoking AI Functions through SQL 
-- https://docs.databricks.com/aws/en/large-language-models/ai-functions

SELECT * FROM contractor_performance;

-- We can use the AI Function to help with trasformations that previously may have ML.
SELECT
  `Incident Report`,
  ai_classify(`Incident Report`, ARRAY('LOW', 'MEDIUM', 'HIGH', 'CRITICAL')) AS incident_serverity
FROM
  contractor_performance
LIMIT 10;

-- AI Extract
SELECT
  `Incident Report`,
  ai_extract(`Incident Report`, ARRAY('LOCATION', 'ASSETS')) AS incident_details
FROM
  contractor_performance
LIMIT 10;

-- AI Query
SELECT * FROM project_master;

SELECT
  ai_query(
    "databricks-claude-sonnet-4",
    CONCAT(
      "You are generating an HTML-formatted project status update email. ",
      "The HTML should be lightweight, clean, and use only the following supported tags: ",
      "<a>, <abbr>, <acronym>, <b>, <blockquote>, <br>, <code>, <div>, <em>, <h1>, <h2>, <h3>, <h4>, <h5>, <h6>, <head>, <hr>, <i>, <li>, <ol>, <p>, <span>, <strong>, <table>, <tbody>, <td>, <th>, <tr>, <ul>. ",
      "Supported attributes: href (for <a>), title (for <a>, <abbr>, <acronym>). Do not include the <html> or <head> tags. ",
      "Structure the summary with a title, a metadata table, and a brief paragraph summarizing the project update. ",
      "Use <strong> for labels, <br> for spacing, and <p> for body text. ",
      "Here is the project information:\n",
      "Project ID: ", `Project ID`, "<br>",
      "Project Name: ", `Project Name`, "<br>",
      "Start Date: ", `Start Date`, "<br>",
      "End Date: ", `End Date`, "<br>",
      "Total Budget (Million $): ", `Total Budget (Million $)`, "<br>",
      "Region: ", `Region`, "<br>",
      "Project Manager: ", `Project Manager`, "<br><br>",
      "Project Update: ", `Project Update`
    )
  ) AS project_summary_email
FROM
  project_master
LIMIT 10;





```
