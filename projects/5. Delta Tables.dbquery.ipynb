```sql
-- All changes to Delta table in Databricks is fully versioned
DESCRIBE HISTORY forecast_snapshots; -- We can select a specific version of the table

SELECT
  *
FROM
  forecast_snapshots VERSION AS OF 0
ORDER BY
  `Snapshot Month` DESC;

-- Or we can restore the table
RESTORE TABLE forecast_snapshots TO VERSION AS OF 0;

-- Delta Tables also indexing through clustering
-- Example: CREATE TABLE table1(col0 INT, col1 string) CLUSTER BY (col0);
ALTER TABLE
  forecast_snapshots
  CLUSTER BY (`Project ID`, `Snapshot Month`);

-- We can also specify automatic cluster column selection.
ALTER TABLE
  forecast_snapshots
  CLUSTER BY AUTO;

-- We can also specify automatic cluster column selection.
ALTER TABLE
  forecast_snapshots
  CLUSTER BY AUTO;

-- Delta tables also support change data capture (CDC)
ALTER TABLE gshen_catalog.enb_projects_workshop.forecast_snapshots SET TBLPROPERTIES (delta.enableChangeDataFeed = true);

-- providing only the startingVersion/timestamp
SELECT * FROM table_changes('gshen_catalog.enb_projects_workshop.forecast_snapshots', 7)

SELECT * FROM table_changes('gshen_catalog.enb_projects_workshop.forecast_snapshots', 7, 10)

-- timestamp as string formatted timestamps
SELECT * FROM table_changes('gshen_catalog.enb_projects_workshop.forecast_snapshots', '2025-07-20 00:51:02')

```
