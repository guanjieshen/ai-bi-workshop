```sql
-- We can use a stored procedure to append a forecast snapshot to the forecast_snapshots table

CREATE OR REPLACE PROCEDURE gshen_catalog.enb_projects_workshop.append_forecast_snapshot (
  IN snapshot_date DATE COMMENT 'Date for the snapshot'
)
LANGUAGE SQL
SQL SECURITY INVOKER
NOT DETERMINISTIC
MODIFIES SQL DATA
COMMENT 'Appends current forecast values into forecast_snapshots with a snapshot date'
AS
BEGIN
  INSERT INTO gshen_catalog.enb_projects_workshop.forecast_snapshots
  SELECT
    `Project ID`,
    snapshot_date AS `Snapshot Month`,
    `Estimate to Complete ($)`,
    `Risk Flag`,
    `Historical Risk Outcome` AS `Historical Risk Flag`,
    null as `_rescued_data`
  FROM gshen_catalog.enb_projects_workshop.forecasts;

END;

-- Invoke the store procedure
CALL gshen_catalog.enb_projects_workshop.append_forecast_snapshot(DATE('2025-07-18'))

SELECT * FROM gshen_catalog.enb_projects_workshop.forecast_snapshots order by `Snapshot Month` desc;

```
