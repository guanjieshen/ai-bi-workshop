```sql
CREATE OR REFRESH MATERIALIZED VIEW  gshen_catalog.enbridge_sr_workshop.incidents_base AS
SELECT
  a.OPERATOR_ID,
  a.operator_name,
  a.incident_year,
  a.REPORT_NUMBER,
  a.state,
  a.LOCATION_LATITUDE,
  a.LOCATION_LONGITUDE,
  a.COMMODITY_RELEASED_TYPE,
  a.SYSTEM_PART_INVOLVED,
  a.ACCIDENT_DETAILS,
  a.barrels_released,
  a.RECOVERED_BBLS,
  a.IGNITE_IND,
  a.EXPLODE_IND,
  a.fatalities,
  a.injuries,
  s.total_miles,
  s.hca_onshore_miles,
  s.hca_offshore_miles
FROM
  gshen_catalog.enbridge_sr_workshop.accidents a
    LEFT JOIN gshen_catalog.enbridge_sr_workshop.annual s
      ON a.OPERATOR_ID = s.OPERATOR_ID
      AND a.incident_year = s.REPORT_YEAR;

SELECT * FROM  gshen_catalog.enbridge_sr_workshop.incidents_base;
```
